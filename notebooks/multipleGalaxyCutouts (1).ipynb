## Creating Galaxy CutOuts within FITSs Files
(aka zooming in on the FITs files)
Within this notebook, there are 2 parts. The first part is the actual cut-out image. The second part shows the original image and where the cut-out was formed. This notebooks loops through FITs images saved within a folder.. 

```python
#importing necessary packages
import numpy as np
import matplotlib.pyplot as plt

import lsst.afw.fits
from astropy.io import fits


from astropy.nddata import Cutout2D
from astropy import units as u

import os
```

```python
#the os.listdir is used to loop through the files within the folder full of sorted galaxies

i=0
for files in os.listdir('/home/emisanchez/WORK/20Galaxy'):
    #writes the path name of file
    fitsPath= '~/WORK/20Galaxy/'+files
    #opens the file
    fitsFile=fits.open(fitsPath)
    data=fitsFile[0].data
    #creates plot title 
    fname='CloseUPGalaxy'+files
    plt.title(fname)
    position = (120, 120)    #pixels
    size = (31, 31)     # pixels 
    #uses the cutout2d object from astropy and the pixel positions and size determined
    cutout = Cutout2D(data, position, size)
    plt.imshow(cutout.data, cmap='gray', origin='lower')
    i=i+1
    plt.figure(i)
   

plt.show()
```

```python
#this shows where the cutout was taken from. There is a box around the central galaxy within the original FITs image. 

x=0
for i in os.listdir('/home/emisanchez/WORK/20Galaxy'):
    fitsPath= '~/WORK/20Galaxy/'+i
    
    fitsFile=fits.open(fitsPath)
    data=fitsFile[0].data
    
    fname='CloseUpGalaxy'+i
    plt.title(fname)
    position = (120, 120)     #pixels
    size = (31, 31)     # pixels 
    #cutout2D from 
    cutout = Cutout2D(data, position, size)
    plt.imshow(data,cmap='gray', origin='lower')
    cutout.plot_on_original(color='white')
    x=x+1
    plt.figure(x)
   

plt.show()
```
