## Display Downloaded FITs Files
The files you display must be in your directory

```python
#import all the packages necessary
import matplotlib.pyplot as plt

import numpy as np 

import lsst.afw.fits
from astropy.io import fits
```

### Run the code below to display one file

```python
#set the fitsPath to the file path
fitsPath= "~/WORK/W2/fitStampSGRB/G1_rings.v3.skycell.0718.018.stk.g.unconv.fits"

#open the fits file
fitsFile=fits.open(fitsPath)

#origin='lower' keeps the image from rotating
#cmap='gray' keeps the color spectrum black, white, and gray
plt.imshow(fitsFile[0].data,cmap='gray', origin='lower')
```

### The code below will display multiple FITS files
This code is specifically divided between the 5 bands of FITS images. Starting at G band, then R, I, Y, and Z. Ultimately this code will display 175 images; as there are 35 galaxy images within each band. 

```python
#gbands
#the for loop will run 35 times to display the 35 images in the g band
for i in range(1,36):
    #the files are named G1, G2, G3 etc. the i variable increases 1,2,3 with every loop so the filename must be changed
    fileName= 'G'+str(i)+'.g.fits'
    #fileName is used to retrieve the file path
    fitsPath= '~/fitStamps/gBand/'+fileName
    
    fitsFile=fits.open(fitsPath)

    #creating name for the image heading
    fname='Galaxy'+str(i)+'.g.fits'
    plt.title(fName)
    plt.imshow(fitsFile[0].data, cmap='gray', origin='lower')
    plt.figure(i+1)
   
plt.show()
```
