## Creating Galaxy CutOuts within FITSs Files
(aka zooming in on the FITs files)
Within this notebook, there are 2 parts. The first part of the code displays the actual cut-out image. The second part shows the original image and where the cut-out was formed. This notebooks loops through FITSs images saved within a folder. 

```python
import numpy as np
import matplotlib.pyplot as plt

import lsst.afw.fits
from astropy.io import fits


from astropy.nddata import Cutout2D
from astropy import units as u

import os
```

```python
#the os.listdir is used to loop through the files within the folder full of sorted galaxies

i=0
for files in os.listdir('/home/emisanchez/WORK/20Galaxy'):
    fitsPath= '~/WORK/20Galaxy/'+files
    
    fitsFile=fits.open(fitsPath)
    data=fitsFile[0].data
    
    fname='CloseUPGalaxy'+files
    plt.title(fname)
    position = (120, 120)
    size = (31, 31)     # pixels 
    cutout = Cutout2D(data, position, size)
    plt.imshow(cutout.data, cmap='gray', origin='lower')
    i=i+1
    plt.figure(i)
   

plt.show()
```

```python
x=0
for i in os.listdir('/home/emisanchez/WORK/20Galaxy'):
    fitsPath= '~/WORK/20Galaxy/'+i
    
    fitsFile=fits.open(fitsPath)
    data=fitsFile[0].data
    
    fname='CloseUPGalaxy'+i
    plt.title(fname)
    position = (120, 120)
    size = (31, 31)     # pixels 
    cutout = Cutout2D(data, position, size)
    plt.imshow(data,cmap='gray', origin='lower')
    cutout.plot_on_original(color='white')
    x=x+1
    plt.figure(x)
   

plt.show()
```
